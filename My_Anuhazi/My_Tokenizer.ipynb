**4.Writing the Flux Training Loop**
The model is trained using Cross-Entropy Loss. It takes the unknown language sequence, predicts the English translation tokens one by one, and minimizes the diference between its predictions and the actual known translation.

```julia
using Flux
using Flux: loss, update!

#Define the training objective
function loss_function(model, src_tokens, tgt_tokens, targets)
    #1. Embed inputs
    src_embedded = model.encoder_embed(src_tokens)
    tgt_embedded = model.decoder_embed(tgt_tokens)

    #2. Pass through Transformer layers
    memory = model.encoder(src_embedded)
    output = model.decoder(tgt_embedded, memory)

    #3. Predict next token probabilites
    predictions = model.projection(output)

    #4. Calculate categorical cross entropy loss against target shifts
    return logitcrossentropy(predictions, targets)
end

# Initialize the optimization algorithm(Adam)
optimizer = Flux.setup(Adam(1e-4), model)

#Main training cycle
epochs = 50

for epoch in 1:epochs
    total_loss = 0.0

    for (src_raw, tgt_raw) in zip(unknown_corpus, english_corpus)
        #Process inputs and shift targets for teacher-forcing training
        src_tokens = encode(encoder_unk, src_raw).input
        tgt_tokens = encode(encoder_eng, tgt_raw).input[:, 1:end-1] # Inputs to decoder
        targets = encode(encoder_eng, tgt_raw).input[:, 2:end] #Expected shifted predictions

        #Calculate gradients amd optimize weights
        loss_val, grads = Flux.withgradient(model) do m
            loss_function(m, src_tokens, tgt_tokens, targets)
        end

        Flux.update!(optimizer, model, grads[1])
        total_loss += loss_val
    end

    if epoch % 10 == 0
        println("Epoch: $epoch | Average Loss: $(total_loss / length(unknown_corpus))")
    end

end
```
