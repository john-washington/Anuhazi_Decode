```julia
using DataFrames
using CSV
using Gadfly
using TextAnalysis
using MLJ
using Chain
using Pipe
using StableRNGs
```

```julia
#import kagglehub
#path = kagglehub.dataset_download("team-ai/spam-text-message-classification")
#print("Path to dataset files: ", path)

df = CSV.read("/home/jzhang/1TB/Keylontic_Teachings/Anuhazi_Decode/SPAM_text_message_Data.csv", DataFrames.DataFrame)
first(df, 10) |> pretty
```

```julia
df = @chain df begin
    DataFrames.transform(:Message =>ByRow(x -> StringDocument(x)) => :Message2)
end
```

```julia
remove_case!.(df[:, :Message2])
prepare!.(df[:, :Message2], strip_html_tags | strip_punctuation| strip_numbers)
stem!.(df[:, :Message2])
```

```julia
crps = Corpus(df[:, :Message2])
```

```julia
update_lexicon!(crps)
```

```julia
m = DocumentTermMatrix(crps)
tfidf_mat = tf_idf(m)
```

```julia
dtm(m)
```

```julia
X, y = tfidf_mat, df[:, :Category]
```

```julia
#Pkg.rm("PyCall")
import MLJDecisionTreeInterface
```

```julia
#Pkg.add("MLJDecisionTreeInterface")
```

```julia
DecisionTreeClassifier = @load DecisionTreeClassifier pkg=DecisionTree
tree_model = DecisionTreeClassifier()
```

```julia
evaluate(tree_model, X, y, 
                resampling=CV(shuffle=true),
                measures=[log_loss, accuracy],
                verbosity=0)
```

```julia
#tree = machine(tree_model, coerce(X, Continuous), coerce(y, Multiclass))
tree = machine(tree_model, X, y)
```

```julia
@doc DecisionTree.DecisionTreeClassifier
```

```julia
@doc DecisionTree.DecisionTreeClassifier
```

```julia
rng = StableRNG(42)
train, test = partition(eachindex(y), 0.85, shuffle=true, rng=rng)
```

```julia
MLJ.fit!(tree , rows=train)
```

```julia
?Tables.table
```

```julia
yhat = MLJ.predict(tree, coerce(X[test, :], Continuous))

#To calculate the log-loss or binary cross entropy of our classification algorithm
println("Log-Loss on the Test Set : $(log_loss(yhat, coerce(y[test], Multiclass)) |> mean)")

#To calculate the accuracy of our classification algorithm
println("Accuracy on the Test Set: $(accuracy(mode.(yhat), coerce(y[test], Multiclass)))")

#To generate the Confusion Matrix of our Classification algorithm
ConfusionMatrix()(mode.(yhat), coerce(y[test], Multiclass))
```
